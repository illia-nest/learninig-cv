### Learn Sobel Filter via CNN

In this notebook we will demonstrate how to train a simple CNN to learn to perform Sobel filtering (first order spatial derivatives). We will also see that a CNN can be thought of as a generic model and can be used to approximate any generic image processing algorithm. This is especially useful if a CNN-based implementation is much faster (usually with the aid of GPU or similar dedicated units) than the original algorithm implementation.

```python
import os
os.environ['KERAS_BACKEND'] = 'torch'    # has to be set before keras is imported

import cv2
import numpy as np
import keras
from keras import Model
from keras.layers import Conv2D, Input
from time import time

from matplotlib import pyplot as plt
plt.rcParams['figure.figsize'] = [15, 10]

# Set the seed for reproducibility (python, numpy and the backend)
keras.utils.set_random_seed(1234578790)
```

#### Dataset

We will try to keep it simple. Therefore, we will use one image (any image will do) and its corresponding Sobel derivative.

```python
# Load image and convert to gray-scale
img = cv2.imread('../lesson_5/data/kodim05.png')
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Select x or y spatial derivative
sobel = 'x'

dx, dy = (1, 0) if sobel == 'x' else (0, 1)
out = cv2.Sobel(img, ddepth=cv2.CV_32F, dx=dx, dy=dy)

plt.subplot(121), plt.imshow(img, cmap='gray')
plt.subplot(122), plt.imshow(out, cmap='gray')
```

```python
print(img[100:103, 100:103])
print(out[100:103, 100:103])
```

#### Data Generator

In this example, we will implement a data generator. So far we have been loading the entire dataset in memory and using it for training. However, in many cases, this is not possible since the dataset is simply too big to fit in memory (this is not the case of this example, though). Therefore, we will need to fetch the data batches from disk as they are requested for training purposes (the so called online fetching). For that, Python provides us with the generator functions.

```python
def datagen(img, out, win=128, batch_size=16):
    rows, cols = img.shape
    
    while True:
        batch_x, batch_y = [], []
        for _ in range(batch_size):
            # Select any random image patch (region)
            r_init = np.random.randint(win, rows-win)
            c_init = np.random.randint(win, cols-win)
            
            # Generate inputs and labels
            x_sample = img[r_init:r_init+win, c_init:c_init+win]
            y_sample = out[r_init:r_init+win, c_init:c_init+win]
            
            # Normalize
            x_sample = (x_sample - np.mean(x_sample))/np.std(x_sample)
            y_sample = (y_sample - np.mean(y_sample))/np.std(y_sample)
            
            # Build batches
            batch_x.append(x_sample)
            batch_y.append(y_sample)
            
        yield np.array(batch_x).astype(np.float32), np.array(batch_y).astype(np.float32)
```

Let's now demonstrate how data generators work.

```python
# Create data generator
dg = datagen(img, out)

# Request next sample
x, y = next(dg)

print(x.shape, y.shape)
```

### Building the Network

The objective is to build a network that will learn to approximate the Sobel operator. The Sobel operator that we use in this example is the classic 3x3 kernel. Therefore, our CNN will consist of one single 3x3 convolutional kernel. And since we are interested in approximating the kernel values, we are not applying any non-linearities to the output.

```python
inputs = Input(shape=(128, 128, 1))
outputs = Conv2D(1, kernel_size=(3, 3), activation="linear", padding='same')(inputs)

model = Model(inputs, outputs)
model.summary()
```

Note that, unlike the classic Sobel filter, the convolutional kernel includes one more parameter: the bias. So the model has 3x3 + 1 = 10 trainable parameters.

Also note that, despite its name, a convolutional layer does not flip the kernel, i.e., it computes a correlation, exactly like `cv2.filter2D` (and `cv2.Sobel`) do. That is why we can later use the learnt weights directly in `cv2.filter2D`.

#### Training

Let's now train the network using the data generator. The objective function is the MSE (since we want the output to approximate the Sobel derivative). Furthermore, note that since we are using data generators, we need to specify the number of steps per epoch (since the generator is basically an infinite loop providing data, the trainer does not know how long one epoch lasts).

```python
epochs = 10
model.compile(loss="mse", optimizer="adam", metrics=["mse"])

start = time()
# The generator already yields random patches, so there is nothing to shuffle
history = model.fit(dg, steps_per_epoch=1000, epochs=epochs, shuffle=False, verbose=2)
print('Elapsed time', time() - start)
```

#### Checking the Learnt Kernel

We are now going to check the value of the kernel that the network has learnt. In order to do so, we are going to extract the weights and the bias from the (only) convolutional kernel.

```python
weights, bias = model.layers[-1].get_weights()
print(weights, bias)
```

Remember that the network does not know that it has to learn the exact Sobel filter (with the exact scale). Since we have normalized the data, the Sobel output and any scaled version of it are equivalent targets. Therefore, in order to better visualize the learnt values, we are going to normalize the kernel so its top-left value is equal to -1 (which is true for Sobel kernels in both x and y direction), and compare it to the true Sobel kernel.

```python
# Remove the unit dimensions
weights = weights.squeeze()

# Normalize
weights = weights/np.abs(weights[0,0])
print(weights.round(2))

# The true Sobel kernel
kx, ky = cv2.getDerivKernels(dx, dy, 3)
print(ky @ kx.T)
```

#### Evaluation

We already know that the kernel approximates quite accurately the Sobel filter. In order to visually demonstrate that, we are going to filter the input image with the learnt kernel and compare it to the original derivative image. Note that the bias is (almost) zero, as it should be: the Sobel filter has no offset.

```python
# Compute the estimated spatial derivative using the trained filter
est = cv2.filter2D(img, ddepth=cv2.CV_32F, kernel=weights)

# Compare the results
plt.subplot(121), plt.imshow(out, cmap='gray'), plt.title('Sobel')
plt.subplot(122), plt.imshow(est, cmap='gray'), plt.title('CNN approximation')

print('Bias:', bias)
print('Correlation:', np.corrcoef(out.flatten(), est.flatten())[0, 1])
```
