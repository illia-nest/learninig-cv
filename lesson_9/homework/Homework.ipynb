### Step 0

Run the necessary imports.

```python
import cv2
import sys
import dlib
import numpy as np
from matplotlib import pyplot as plt
import matplotlib.patches as patches
plt.rcParams['figure.figsize'] = [15, 10]
```

### Step 1

Load an image (any image that contains faces).

```python
# Although the original image is gray, 
# in fact, it is still read as a 3-channel picture, 
# so we need to convert it to grayscale anyway.
img = cv2.imread('../data/Kuchma_in_Kyiv.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY) 
plt.imshow(img)
```

### Step 2

Load the dlib face predictor.

```python
# Loading basic HOG detector from dlib
detector = dlib.get_frontal_face_detector()
```

### Step 3

Run the predictor on your image.

```python
# Detecting fac
# es in the grayscale image
detected_faces = detector(gray, 1)
```

### Step 4

Draw bounding boxes around the detected faces and plot the image. Use different colour for each face.

```python
# Define palette for future viz
colors = [
    "#E63946",  # red
    "#F4A261",  # orange
    "#E9C46A",  # yellow
    "#2A9D8F",  # teal
    "#00B4D8",  # cyan
    "#457B9D",  # blue
    "#6C5CE7",  # purple
    "#D45087",  # pink
    "#7A9E45",  # green
    "#8D6E63",  # brown
]

fig, ax = plt.subplots()
ax.imshow(img)

# Looping through found faces and draw rectangles using matplotlib patches
for idx, face in enumerate(detected_faces):
    # Getting coordinates from dlib rectangle object
    x = face.left()
    y = face.top()
    w = face.width()
    h = face.height()
    
    # Creating a matplotlib rectangle patch (linewidth, edgecolor, and facecolor)
    rect = patches.Rectangle(
        (x, y), w, h, 
        linewidth=2, 
        edgecolor=colors[idx], 
        facecolor='none'
    )
    
    # Adding the patch to the axes
    ax.add_patch(rect)

# Hiding axes tick marks and show the plot
plt.axis('off')
plt.title(f"Detected {len(detected_faces)} Face(s)")
plt.show()
```
