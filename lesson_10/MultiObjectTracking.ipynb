## Multi-Object Tracking

In `Tracking.ipynb` we followed **one** object that we selected by hand. In most applications (traffic, surveillance, sports, ...) there are **many** objects, they appear and disappear, and nobody tells the tracker where they are. The standard solution is **tracking by detection**:

1. A **detector** finds all objects of interest in every frame (lesson 18). It gives boxes, but no identities.
2. A **tracker** links the boxes of consecutive frames, so that every physical object keeps **its own ID**: it **predicts** where each known object should be now (the *motion model*, a Kalman filter), **associates** the predictions with the new detections (usually by their overlap, solved with the Hungarian algorithm), and then updates the known objects, creates new ones for unmatched detections and deletes the objects that have been lost for too long.

The tracker never looks at the image itself (except for the camera-motion estimate of BoT-SORT that we meet at the end), so the same tracker can be combined with any detector. In this notebook we use three libraries:

* [`ultralytics`](https://github.com/ultralytics/ultralytics): the YOLOv8 detector from lesson 18,
* [`supervision`](https://github.com/roboflow/supervision): a toolbox for detections and their visualisation,
* [`trackers`](https://github.com/roboflow/trackers): ready-made implementations of the SORT, ByteTrack, OC-SORT and BoT-SORT trackers.

### Data

We use a video of the *Patrouille de France* aerobatic team rehearsing: four jets that fly in tight formation, split, cross each other and fly in front of a moving camera. It is a public-domain video from [Wikimedia Commons](https://commons.wikimedia.org/wiki/File:Patrouille_de_France_Demonstration.webm). Download it and place it as `data/patrouille_de_france.webm`.

````python
import cv2
import numpy as np
import supervision as sv
from IPython.display import Video
from matplotlib import pyplot as plt
from tqdm.auto import tqdm
from trackers import SORTTracker, ByteTrackTracker, OCSORTTracker, BoTSORTTracker
from ultralytics import YOLO
plt.rcParams['figure.figsize'] = [15, 10]
````

We will work with two short clips of the video:

* **Clip A**: the four jets fly in a close formation (easy: the aircraft are large and all of them are detected).
* **Clip B**: the jets split up, they are small and their trails confuse the detector (hard).

````python
video_path = 'data/patrouille_de_france.webm'
info = sv.VideoInfo.from_video_path(video_path)
print(info)

clips = {'A': (33, 41), 'B': (64.5, 72)}    # start and end of the clips in seconds


def read_frames(clip):
    start, end = clips[clip]
    return sv.get_video_frames_generator(video_path, start=int(start * info.fps), end=int(end * info.fps))
````

````python
frame = next(read_frames('A'))
plt.subplot(121), plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)), plt.title('Clip A')
frame = next(read_frames('B'))
plt.subplot(122), plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)), plt.title('Clip B')
````

## 1. Detection

We use the YOLOv8 model from lesson 18, pretrained on COCO. COCO has an `airplane` class, so we do not need to train anything. The model runs on the GPU if there is one and on the CPU otherwise (the CPU is slower, but still fine for these short clips).

The model returns a result for every frame; `sv.Detections.from_ultralytics` converts it to the `sv.Detections` object that `supervision` and the trackers work with (boxes, confidences, class ids). We keep only the aircraft.

We deliberately use a **low confidence threshold** (0.1) and keep the weak detections too. We will see in a moment why.

````python
model = YOLO('yolov8s.pt')
AIRPLANE = 4    # class id in COCO


def detect(clip, conf=0.1):
    detections = []
    start, end = clips[clip]
    for frame in tqdm(read_frames(clip), total=int((end - start) * info.fps)):
        result = model.predict(frame, conf=conf, verbose=False)[0]
        det = sv.Detections.from_ultralytics(result)
        detections.append(det[det.class_id == AIRPLANE])
    return detections
````

````python
detections = {clip: detect(clip) for clip in clips}
````

The detector is the slow part, so we run it **only once** and keep the results. Every tracker below will work with these stored detections, which also guarantees that we compare the trackers on exactly the same input.

Let's look at one frame. Every box has a confidence, but no identity.

````python
frame = next(read_frames('A'))
det = detections['A'][0]
frame = sv.BoxAnnotator(thickness=3).annotate(frame, det)
frame = sv.LabelAnnotator(text_scale=1, text_thickness=2).annotate(frame, det, labels=[f'{c:.2f}' for c in det.confidence])
plt.imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
````

How well does the detector alone do? In both clips there are always **four** jets. This is the number of detections per frame:

````python
for i, (clip, dets) in enumerate(detections.items()):
    plt.subplot(1, 2, i + 1)
    plt.plot([len(d) for d in dets], label='confidence >= 0.1')
    plt.plot([(d.confidence >= 0.5).sum() for d in dets], label='confidence >= 0.5')
    plt.title(f'Clip {clip}'), plt.xlabel('Frame'), plt.ylabel('Detections'), plt.ylim(0, 8), plt.legend()
````

In clip A the detector is almost perfect: it finds all four jets in every frame (only a few frames have a spurious fifth box). In clip B the number of detections jumps between 0 and 8, and with a high confidence threshold the detector often finds only two jets or none. But even in clip A we get *boxes*, not *objects*: nothing tells us which box in frame `t+1` is the same jet as which box in frame `t`.

## 2. Tracking with ByteTrack

[ByteTrack](https://arxiv.org/abs/2110.06864) is a widely used tracker. It extends the classic SORT tracker by one simple idea (step 3 below). It keeps a list of **tracks**, one for every object it believes to be in the scene, and processes every frame in four steps:

1. **Predict.** Every track has its own **Kalman filter**, the motion model from the lecture. From the recent history of the box (its position and speed) it predicts where the box should be in the new frame. The filter also knows how uncertain its prediction is, which tells it how much to trust the prediction and how much to trust the new detection (slide 13). Thanks to the prediction, a track keeps moving along its path even in a frame in which the detector misses the object.
2. **Associate.** Which detection belongs to which track? Every predicted box is compared with every detected box by their overlap, the **IoU** (1 means identical boxes, 0 means no overlap). The **Hungarian algorithm** then finds the assignment of the detections to the tracks with the best total overlap, and pairs with too little overlap are rejected. In this step ByteTrack uses only the *confident* detections.
3. **Second chance (the idea of ByteTrack).** Most trackers throw away the detections with a low confidence. But a low confidence often means that the object is *partly hidden or blurred*, not that it is not there. So the tracks that found no partner in step 2 get a second chance: they are matched with the **low-confidence** detections. This recovers objects that are hard to see, while the weak detections can never start a new track by themselves.
4. **Update the tracks.** A matched track is corrected by its detection (the Kalman filter combines the prediction and the measurement). An unmatched confident detection starts a **new track**, which becomes confirmed (gets an ID) only after it has been matched in two consecutive frames, so a single false alarm does not create an ID. A track without a match is kept as **lost** for a while (30 frames by default), so the object gets its old ID back if it reappears; after that, the track is deleted.

Note that the tracker never looks at the image, only at the boxes: ByteTrack has a **motion model** (the Kalman filter) but no **appearance model** (slide 12).

All trackers of the `trackers` package have the same interface: we create the tracker and call `update` with the detections of every frame. It returns the same detections with an added `tracker_id`. An ID of `-1` means that the track is not confirmed yet: by default a new object has to be matched in two consecutive frames before it gets an ID.

````python
def track(tracker, dets, frames=None):
    tracked = []
    for i, det in enumerate(dets):
        out = tracker.update(det) if frames is None else tracker.update(det, frame=frames[i])
        tracked.append(out[out.tracker_id >= 0])
    return tracked


def n_ids(tracked):
    return len(set(np.concatenate([t.tracker_id for t in tracked])))


tracked_a = track(ByteTrackTracker(frame_rate=info.fps), detections['A'])
print('Number of IDs given by ByteTrack in clip A:', n_ids(tracked_a), '(there are 4 jets)')
````

### Watch the result

`supervision` has annotators that draw the boxes, the ID labels and the **traces** (the recent trajectory of every ID). We color everything by the tracker ID, so an object keeps its color as long as it keeps its ID. We save the result as a video (VP8 codec, see `Tracking.ipynb`) at half the resolution and embed it in the notebook.

````python
def make_annotator():
    box_annotator = sv.BoxAnnotator(thickness=3, color_lookup=sv.ColorLookup.TRACK)
    label_annotator = sv.LabelAnnotator(text_scale=1, text_thickness=2, color_lookup=sv.ColorLookup.TRACK)
    trace_annotator = sv.TraceAnnotator(thickness=3, trace_length=45, color_lookup=sv.ColorLookup.TRACK)

    def annotate(frame, det):
        frame = trace_annotator.annotate(frame, det)
        frame = box_annotator.annotate(frame, det)
        return label_annotator.annotate(frame, det, labels=[f'#{i}' for i in det.tracker_id])
    return annotate


def render(clip, tracked, path):
    annotate = make_annotator()
    size = (info.width // 2, info.height // 2)
    writer = cv2.VideoWriter(path, cv2.VideoWriter_fourcc(*'VP80'), info.fps, size)
    for frame, det in zip(read_frames(clip), tracked):
        writer.write(cv2.resize(annotate(frame, det), size))
    writer.release()
    return Video(path, embed=True, html_attributes='controls loop')
````

````python
render('A', tracked_a, 'data/bytetrack_a.webm')
````

## 3. A harder case

Clip B is much harder: the jets are small, they cross the picture while the camera follows them, and the detector misses some of them in many frames. Let's run ByteTrack on it and look.

````python
tracked_b = track(ByteTrackTracker(frame_rate=info.fps), detections['B'])
print('Number of IDs given by ByteTrack in clip B:', n_ids(tracked_b), '(there are 4 jets)')
render('B', tracked_b, 'data/bytetrack_b.webm')
````

Even ByteTrack needs more than four IDs here: tracks are broken when the detector misses a jet for a longer time, and the new detection of the same jet then gets a new ID. Towards the end of the clip the jets are so small that the detector finds nothing at all, and a tracker cannot follow what is never detected.

## 4. Comparing trackers

Since all trackers share the same interface, comparing them is easy. We run four trackers on the same stored detections of clip B:

| Tracker | Idea |
|---|---|
| **SORT** | The classic baseline: Kalman filter + Hungarian matching of the boxes by their overlap (IoU) |
| **ByteTrack** | SORT + the second association stage with the low-confidence detections |
| **OC-SORT** | SORT that corrects the Kalman filter by the real observations after an object has been lost, and uses the direction of motion in the association |
| **BoT-SORT** | ByteTrack-like association + **camera motion compensation**. It needs the image (to estimate the camera motion), so we pass the frames too |

We have no ground truth here, but we know that there are four jets. A good tracker gives **each jet one ID for the whole clip**, so the total number of IDs should stay close to four. Every additional ID means that a track has been broken.

````python
all_trackers = {'SORT': SORTTracker(frame_rate=info.fps),
            'ByteTrack': ByteTrackTracker(frame_rate=info.fps),
            'OC-SORT': OCSORTTracker(frame_rate=info.fps),
            'BoT-SORT': BoTSORTTracker(frame_rate=info.fps)}

frames_b = list(read_frames('B'))
results = {name: track(tracker, detections['B'], frames_b if name == 'BoT-SORT' else None)
           for name, tracker in all_trackers.items()}

for name, tracked in results.items():
    print(f'{name:10s} {n_ids(tracked):3d} IDs')
````

An **ID timeline** shows what happens: every dot is one tracked box, its horizontal position is the frame and its vertical position is the ID. A jet that is tracked well is one long horizontal line. A broken track shows up as a line that stops and a new line (with a new ID) that starts.

````python
for i, (name, tracked) in enumerate(results.items()):
    plt.subplot(len(results), 1, i + 1)
    for frame_idx, det in enumerate(tracked):
        plt.scatter([frame_idx] * len(det), det.tracker_id, c='k', s=4)
    plt.title(f'{name}: {n_ids(tracked)} IDs'), plt.ylabel('Tracker ID')
plt.xlabel('Frame')
plt.tight_layout()
````

SORT breaks the tracks much more often than the other trackers: it gives the jets 21 IDs, OC-SORT 16, while ByteTrack and BoT-SORT stay at 6 and 7. In the timeline, the IDs of SORT and OC-SORT keep growing, whereas ByteTrack and BoT-SORT stay on a few low IDs. The camera motion compensation of BoT-SORT does not bring an improvement over ByteTrack in this clip (7 vs. 6 IDs).

### Watching the trackers side by side

Numbers and timelines are abstract, so let's **watch** the four trackers work on the same clip. Every panel of the video is one tracker, and all of them get exactly the same detections. The counter in each panel shows **how many different IDs the tracker has handed out so far**. There are only four jets, so a counter that keeps growing means that the tracker keeps losing the jets and starting new tracks. Watch the ID labels (and colors) on a single jet: does it keep its ID?

````python
def render_grid(clip, results, path):
    names = list(results)
    annotators = {name: make_annotator() for name in names}
    seen = {name: set() for name in names}

    size = (info.width // 2, info.height // 2)    # size of one panel
    writer = cv2.VideoWriter(path, cv2.VideoWriter_fourcc(*'VP80'), info.fps, (2 * size[0], 2 * size[1]))
    for i, frame in enumerate(read_frames(clip)):
        panels = []
        for name in names:
            det = results[name][i]
            seen[name].update(det.tracker_id.tolist())
            panel = cv2.resize(annotators[name](frame, det), size)

            text = f'{name}: {len(seen[name])} IDs so far'
            cv2.putText(panel, text, (15, 40), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 0), 5)
            cv2.putText(panel, text, (15, 40), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
            panels.append(panel)
        writer.write(np.vstack([np.hstack(panels[:2]), np.hstack(panels[2:])]))
    writer.release()
    return Video(path, embed=True, html_attributes='controls loop')
````

````python
render_grid('B', results, 'data/comparison_b.webm')
````

### Why does the confidence threshold matter?

At the beginning we kept the detections with a confidence as low as 0.1. SORT matches **all** the detections it gets, so a weak, wrong detection easily starts a new track. ByteTrack uses the weak detections only in the second stage, to prolong existing tracks, so it can profit from them. Let's feed both trackers with different minimum confidences.

````python
for conf in [0.1, 0.25, 0.5]:
    dets = [d[d.confidence >= conf] for d in detections['B']]
    sort_ids = n_ids(track(SORTTracker(frame_rate=info.fps), dets))
    byte_ids = n_ids(track(ByteTrackTracker(frame_rate=info.fps), dets))
    print(f'Minimum confidence {conf:4.2f}:  SORT {sort_ids:3d} IDs,  ByteTrack {byte_ids:3d} IDs')
````

With the weak detections removed (minimum confidence 0.5), SORT gets better (fewer IDs), but ByteTrack gets *worse*: it relies on the weak detections to bridge the gaps. With all the detections, ByteTrack is by far the better of the two.

## Summary

* **Tracking by detection**: the detector finds the objects in every frame, the tracker gives them persistent identities (slide 11: *identity preservation*). The motion model is a Kalman filter (slide 13), and the association is an optimisation: the best assignment of the detections to the tracks.
* All these trackers use the **same simple interface**, so they can be exchanged easily and combined with any detector.
* The tracker is only as good as the detector: **ID switches and broken tracks** appear where the detections are missing, and the choice of the tracker and its settings changes how well it copes.
* The `trackers` package offers more settings (e.g. `lost_track_buffer`, how long a lost object is remembered; `minimum_consecutive_frames`, how long an object has to be seen before it gets an ID). Try them on clip B.
* Trackers that also use the **appearance** of the objects (re-identification networks, e.g. DeepSORT) are more robust to long occlusions, at the price of an additional network. In this video all the jets look the same, so the appearance would not help much.
