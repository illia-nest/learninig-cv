```python
import time
from pathlib import  Path
import cv2
import numpy as np
from IPython.display import display, Image, clear_output
from matplotlib import pyplot as plt
plt.rcParams['figure.figsize'] = [15, 10]
```

```python
def draw_box(img, box, color=(0, 255, 0), thickness=2):
    x, y, w, h = [int(v) for v in box]
    return cv2.rectangle(img.copy(), (x, y), (x + w, y + h), color, thickness)
```

### Step 1
Decide what video you are going to use for this homework, select an object and generate the template. You can use any video you want (your own, from Youtube, etc.)
and track any object you want (e.g. a car, a pedestrian, etc.).


```python
cap = cv2.VideoCapture('./bottle.mp4')
ret, frame = cap.read()
assert ret

bbox =  (1060, 390, 130, 275)

src_color = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
src_color_painted = draw_box(src_color, bbox)
plt.imshow(src_color_painted)
```

### Step 2
Initialize a tracker (e.g. KCF).

```python
tracker = cv2.TrackerKCF_create()
ok = tracker.init(frame, bbox)
```

### Step 3
Run the tracker on the video and the selected object. Run the tracker for around 10-15 frames.

### Step 4
For each frame, print the bounding box on the image and save it.

```python
MAX_FRAMES = 25  # Maximum number of frames to process
frame_count = 0

# Process video
while cap.isOpened() and frame_count < MAX_FRAMES:
    ret, frame = cap.read()
    if not ret:
        break

    # Update tracker
    success, bbox = tracker.update(frame)

    if success:
        x, y, w, h = map(int, bbox)

        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        cv2.putText(
            frame,
            "Tracking",
            (x, max(20, y - 10)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )
    else:
        cv2.putText(
            frame,
            "Tracking lost",
            (30, 50),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 0, 255),
            2
        )

    # Display frame in Jupyter Notebook
    _, buffer = cv2.imencode(".jpg", frame)
    
    # filename = Path("./kcf_results") / f"frame_{frame_count:05d}.jpg"
    # filename.write_bytes(buffer.tobytes())

    clear_output(wait=True)
    display(Image(data=buffer.tobytes()))

    frame_count += 1
    time.sleep(0.1)

cap.release()
print(f"Processed {frame_count} frames.")
```

### Step 5
Select a different tracker (e.g. CSRT) and repeat steps 2, 3 and 4.

```python
cap = cv2.VideoCapture('./bottle.mp4')
ret, frame = cap.read()
assert ret

bbox =  (1060, 390, 130, 275)

tracker = cv2.TrackerCSRT_create()
ok = tracker.init(frame, bbox)

MAX_FRAMES = 25  # Maximum number of frames to process
frame_count = 0

# Process video
while cap.isOpened() and frame_count < MAX_FRAMES:
    ret, frame = cap.read()
    if not ret:
        break

    # Update tracker
    success, bbox = tracker.update(frame)

    if success:
        x, y, w, h = map(int, bbox)

        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        cv2.putText(
            frame,
            "Tracking",
            (x, max(20, y - 10)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )
    else:
        cv2.putText(
            frame,
            "Tracking lost",
            (30, 50),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 0, 255),
            2
        )

    # Display frame in Jupyter Notebook
    _, buffer = cv2.imencode(".jpg", frame)
    
    # filename = Path("./csrt_results") / f"frame_{frame_count:05d}.jpg"
    # filename.write_bytes(buffer.tobytes())

    clear_output(wait=True)
    display(Image(data=buffer.tobytes()))

    frame_count += 1
    time.sleep(0.1)

cap.release()
print(f"Processed {frame_count} frames.")
```
